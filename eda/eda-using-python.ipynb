```python
bins = [0, 100000, 300000, float('inf')]
lables = ['Low', 'Medium', 'High']
df['Income Band'] = pd.cut(df['Estimated Income'], bins=bins, labels=lables, right=False)
df.head()
```

```python
df['Income Band'].value_counts().plot(kind='bar')
```

```python
categorical_cols = [
    "BRId", "IAId", "Amount of Credit Cards",
    "Nationality", "Occupation", "Fee Structure",
    "Loyalty Classification", "Properties Owned",
    "Risk Weighting", "Income Band"

]

for col in categorical_cols:
    print(f"Value counts for '{col}':")
    display(df[col].value_counts().plot(kind='bar'))
    print("\n")

```

## Univariate Analysis

```python
for i, predictor in enumerate (df[["BRId", "IAId", "Amount of Credit Cards",
    "Nationality", "Occupation", "Fee Structure",
    "Loyalty Classification", "Properties Owned",
    "Risk Weighting", "Income Band"]]):
  plt.figure(i)
  sns.countplot(data=df , x=predictor)

```

## Bivariant Analysis



```python
for i, predictor in enumerate (df[["BRId", "IAId", "Amount of Credit Cards",
    "Nationality", "Occupation", "Fee Structure",
    "Loyalty Classification", "Properties Owned",
    "Risk Weighting", "Income Band"]]):
  plt.figure(i)
  sns.countplot(data=df , x=predictor, hue='Nationality')
```

```python
for col in categorical_cols:
    if col == "Occupation":
        continue #skip the Occupation
    plt.figure(figsize=(8,4))# 8 inches and height of 4 inches.
    sns.histplot(df[col])
    plt.title('Histogram of Occupation Count')
    plt.xlabel(col)
    plt.ylabel("Count")
    plt.show()
```

```python
numerical_cols = ['Estimated Income','Superannuation Savings','Credit Card Balance','Bank Loans','Bank Deposits','Checking Accounts','Saving Accounts','Foreign Currency Account','Business Lending']
# univariate analysis and visualization
plt.figure(figsize =(20,15))
for i ,col in enumerate(numerical_cols):
  plt.subplot(5,3,i+1)
  sns.histplot(df[col],kde=True)
plt.show()

```

## Heatmaps

```python
numerical_cols = ['Estimated Income','Superannuation Savings','Credit Card Balance','Bank Loans','Bank Deposits','Checking Accounts','Saving Accounts','Foreign Currency Account','Business Lending']

correlation_matrix = df[numerical_cols].corr() #Produces a square table (matrix) where:                                                                                                                           #Values are between -1 and 1
plt.figure(figsize=(10,8))
sns.heatmap(correlation_matrix, annot=True, cmap='crest', fmt=".2f") # Draws a colored grid (heatmap) showing correlation values, annot=true Writes the correlation numbers on each cell.
plt.title("correlation matrix")#cmap='crest' → Sets the color palette (in this case, “crest” gives a blue‑green gradient).,fmt=".2f" → Formats numbers to 2 decimal places (e.g., 0.78 instead of 0.782341).

plt.show()
```
